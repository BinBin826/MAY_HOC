## Bài 7


```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean dim=0:", torch.mean(a, dim=0))
print("\nStd dim=0:", torch.std(a, dim=0))
print("\nCumsum dim=0:\n", torch.cumsum(a, dim=0))
print("\nMean dim=1:", torch.mean(a, dim=1))
print("\nStd dim=1:", torch.std(a, dim=1))
print("\nCumsum dim=1:\n", torch.cumsum(a, dim=1))
```

**Nhận xét / trả lời yêu cầu:**

dim=0 tính theo chiều đầu tiên, nên mean/std cho từng cột và cumsum cộng dồn theo cột. dim=1 tính theo chiều thứ hai, nên mean/std cho từng hàng và cumsum cộng dồn theo hàng.



## Bài 8


```python
import torch
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3*x2**2 + 2*x2 + 1
y2.backward()
print("\nFor y = 3*x**2 + 2*x + 1:")
print("y =", y2)
print("x.grad =", x2.grad)
```

**Nhận xét / trả lời yêu cầu:**

Với y=x², dy/dx=2x; tại x=3.6 được 7.2, đúng với x.grad. Với y=3x²+2x+1, dy/dx=6x+2; tại x=3.6 được 23.6, đúng với autograd.



## Bài 9


```python
import numpy as np
import torch
np.random.seed(42)
torch.manual_seed(42)
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
print("Final MSE:", mse.item())
```

**Nhận xét / trả lời yêu cầu:**

Hệ số học được xấp xỉ [1, 2, 3], đúng với đa thức ban đầu y=x²+2x+3. zero_grad() xóa gradient cũ; backward() tính gradient của MSE; step() cập nhật hệ số theo optimizer.



## Bài 10


```python
import random
import torch
random.seed(42)
torch.manual_seed(42)
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(2000):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())
print("\nCheck:")
print("A+B =", (A+B).item())
print("C-D =", (C-D).item())
print("A+C =", (A+C).item())
print("B-D =", (B-D).item())
```

**Nhận xét / trả lời yêu cầu:**

Nghiệm thỏa bốn phương trình với sai số nhỏ. Bài toán có vô số nghiệm vì một phương trình phụ thuộc vào ba phương trình còn lại. Đặt A=t thì B=9−t, C=8−t, D=7−t. Optimizer tìm một nghiệm cụ thể phụ thuộc giá trị khởi tạo.



## Bài 11


```python
import numpy as np
import torch

X = torch.arange(1, 25)
print("4 x 6:\n", X.reshape(4, 6))
print("2 x 12:\n", X.reshape(2, 12))
print("2 x 3 x 4:\n", X.reshape(2, 3, 4))

torch.manual_seed(42)
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("\nA+B:\n", A+B)
print("\nA*B:\n", A*B)
print("\nA/B:\n", A/B)
print("\nMean by column:", torch.mean(A, dim=0))
print("\nStd by column:", torch.std(A, dim=0))

x = torch.tensor(2.0, requires_grad=True)
y = 2*x**2 + 5*x + 3
y.backward()
print("\ny =", y.item())
print("dy/dx by autograd =", x.grad.item())
print("dy/dx manual =", 13)

np.random.seed(42)
torch.manual_seed(42)
X_np = np.random.randn(20, 1) * 5
Y_np = 2*X_np**2 - 3*X_np + 5
XX = np.hstack([X_np**2, X_np, np.ones_like(X_np)])
w = torch.randn(3, 1, requires_grad=True)
x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y_np, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
for epoch in range(1500):
    optimizer.zero_grad()
    pred = x_train @ w
    loss = torch.mean((y_train - pred)**2)
    loss.backward()
    optimizer.step()
print("\nLearned coefficients:")
print(w)
print("Final MSE:", loss.item())
```

**Nhận xét / trả lời yêu cầu:**

X có 24 phần tử nên cả ba phép reshape đều hợp lệ. A+B, A*B, A/B là phép toán theo phần tử; mean và std theo cột dùng dim=0. Với y=2x²+5x+3, dy/dx=4x+5 nên tại x=2 là 13, đúng với autograd. Các hệ số học được xấp xỉ [2, −3, 5].
